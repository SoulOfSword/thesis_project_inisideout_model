# Model Comparison: Inside-Out vs Non-Inside-Out

```python
import numpy as np

io_data = np.load('../outputs/mcmc_results/bic_data_io_model.npz')
non_io_data = np.load('../outputs/mcmc_results/bic_data_non_io_model.npz')

print("Inside-Out Model:")
print(f"  Best params: n = {io_data['best_n']:.4f}, k = {io_data['best_k']:.4f}")
print(f"  log(L_max) = {io_data['log_L_max']:.2f}")
print(f"  f_gas chi2 = {io_data['chi2_fgas']:.2f}")
if 'n_clipped' in io_data:
    print(f"  Clipped galaxies: {io_data['n_clipped']}/117")

print("\nNon-Inside-Out Model:")
print(f"  Best params: a = {non_io_data['best_a']:.4f}, b = {non_io_data['best_b']:.4f}")
print(f"  log(L_max) = {non_io_data['log_L_max']:.2f}")
print(f"  f_gas chi2 = {non_io_data['chi2_fgas']:.2f}")
```

```python
n_data = 117
k = 2

BIC_io = k * np.log(n_data) - 2 * io_data['log_L_max']
BIC_non_io = k * np.log(n_data) - 2 * non_io_data['log_L_max']

print(f"BIC (Inside-Out):     {BIC_io:.2f}")
print(f"BIC (Non-Inside-Out): {BIC_non_io:.2f}")
print(f"(IO - Non-IO):   {BIC_io - BIC_non_io:.2f}")
```

```python
chi2_io = io_data['chi2_fgas']
chi2_non_io = non_io_data['chi2_fgas']
dof = n_data - k

print(f"chi2 (Inside-Out):      {chi2_io:.2f}")
print(f"chi2 (Non-Inside-Out):  {chi2_non_io:.2f}")
print(f"(IO - Non-IO):    {chi2_io - chi2_non_io:.2f}")
print(f"\nReduced chi2 (chi2/dof, dof={dof}):")
print(f"  IO:     {chi2_io/dof:.3f}")
print(f"  Non-IO: {chi2_non_io/dof:.3f}")
```
